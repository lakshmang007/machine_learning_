```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

x = np.random.rand(100)
Xtr, Xte = x[:50].reshape(-1,1), x[50:].reshape(-1,1)
ytr = np.where(x[:50] <= 0.5, 'Class1', 'Class2')

plt.figure(figsize=(12,8))

for i,k in enumerate([1,2,3,4,5,20,30],1):
    pred = KNeighborsClassifier(k).fit(Xtr,ytr).predict(Xte)

    print(f"\nk={k}")
    for v,c in zip(Xte[:10], pred[:10]):
        print(round(v[0],4), "->", c)

    plt.subplot(3,3,i)
    plt.scatter(Xtr, np.zeros(50), c=(ytr=='Class2'))
    plt.scatter(Xte, np.ones(50), c=(pred=='Class2'), marker='x')
    plt.title(f'k={k}')

plt.tight_layout()
plt.show()
```
