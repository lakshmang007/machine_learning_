```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("tips.csv")

x = data["total_bill"].values
y = data["tip"].values

def lwr(x0, t=5):
    w = np.exp(-(x-x0)**2/(2*t*t))
    return np.sum(w*y)/np.sum(w)

yp = [lwr(i) for i in x]

plt.scatter(x, y, label="Data")
plt.plot(x, yp, 'r', label="LWR")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.legend()
plt.show()
```
