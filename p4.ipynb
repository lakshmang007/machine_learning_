```python
import pandas as pd
data = pd.read_csv("trainingdata.csv")
X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values
h = ['0'] * len(X[0])
for i in range(len(X)):
    if y[i] == "Yes":
        for j in range(len(h)):
            if h[j] == '0':
                h[j] = X[i][j]
            elif h[j] != X[i][j]:
                h[j] = '?'
        print("Step", i+1, ":", h)
print("\nFinal Hypothesis:", h)
```
